# Welcome to Introduction to Geospatial Python

Through interactive lessons and hands-on exercises, this course introduces you to geographic data analysis using the Python programming language. This course is an adaptation of the course [Automating GIS Processes II](https://autogis-site.readthedocs.io) offered every autumn at the University of Helsinki. The learning materials have been modified for self-learning.

The learning materials are provided and maintained by [Kamyar Hasanzadeh](https://www.helsinki.fi/fi/tutustu-meihin/ihmiset/henkilohaku/kamyar-hasanzadeh-9481798) from University of Helsinki. Samantha Wittke from CSC has also contributed to the preparation of the learning materials for this self-learning course.

## Course format and topics

This course can be fully completed in the CSC Notebook environment, which uses JupyterLab and has all the necessary libraries installed. The course consists of interactive lessons focusing on developing basic programming skills using Python and applying those skills to manipulate and analyze geographic information. The lessons are independent of each other, so you can complete them in any order you like. However, they are arranged in the order we recommend for your learning. The practical examples in this course involve real-world scenarios and data, making it easier to apply the lessons to your own work.

During this course, you will dive into manipulating and analyzing geographic data in Python. This course builds upon the skills introduced in an earlier course offered at the University of Helsinki, [GeoPython](https://geo-python-site.readthedocs.io/), which focuses on learning the basics of Python programming. Thus, the assumption is that you are already familiar with the basics of Python, the Pandas library, and making simple visualizations with libraries such as Matplotlib. As skills in data handling and processing with the Pandas library are essential for a smooth experience in this course, the course starts with Lesson 0, which serves as a quick overview for working with the Pandas library. If you are new to Pandas, we encourage you to check out the relevant [GeoPython](https://geo-python-site.readthedocs.io/) course pages or other learning materials before starting this course.


| Lesson | Theme                                                                        |

| 5      | Visualisation: static and interactive maps                                   |
| 6      | OpenStreetMap data (osmnx) and Network analysis (networkx)                   |
| 7      | Working with Raster data in Python                                           |
